# Editing items in a List

```python
L = [1,2,3,4,5]

# editing with indexing
L[-1] = 500

# editing with slicing
L[1:4] = [200,300,400]

print(L)
```

# Deleting Items From a List

```python
# del
L = [1,2,3,4,5]

# indexing
del L[-1]
del(L)
# slicing
del L[1:3]
print(L)
```

```python
# remove : useful as here we dont need to remember the index of item we want to delete just use the value to be deleted

L = [1,2,3,4,5]

L.remove(5)

print(L)
```

```python
# pop : it removes the item using index ; if you dont provide index by default the last elements will be removed
L = [1,2,3,4,5]

L.pop(0)

print(L)
```

```python
# clear : clears all the list
L = [1,2,3,4,5]

L.clear()

print(L)
```

## List Comprehension

List Comprehension provides a concise way of creating lists.

newlist = [expression for item in iterable if condition == True]image.png

**Expression :** what you want to evaluate or want to print from

**Item :** Most of the cases 'i' that is the element

**Iterable :** What is must loop through

Advantages of List Comprehension
- More time-efficient and space-efficient than loops.
- Require fewer lines of code.
- Transforms iterative statement into a formula.

```python
# Add 1 to 10 numbers to a list
L = []

for i in range(1,11):
  L.append(i)


print(L)
```

```python
L = [i for i in range(1,11)]
print(L)
```

```python
# scalar multiplication on a vector
v = [2,3,4]
s = -3
# [-6,-9,-12]

x = []

for i in v :
  x.append(i*s)

print (x)

[s*i for i in v]
```

```python
# Add squares
L = [1,2,3,4,5]

[i**2 for i in L]

for i in L :
  square = (i**2)
  L[i-1] = square
  square = 0
print(L)
```

```python
# Print all numbers divisible by 5 in the range of 1 to 50

# [i for i in range(1,51) if i%5 == 0]


[i for i in range(5,51,5)]
```

```python
# find languages which start with letter p
languages = ['java','python','php','c','javascript']

# [language for language in languages if language.startswith('p')]
[i for i in languages if i.startswith('p')]
```

```python
 # Nested if with List Comprehension
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']

# add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'

[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
```

```python
# Print a (3,3) matrix using list comprehension -> Nested List comprehension
# A list comprehension inside another list comprehension ([[...]]) → The inner comprehension creates a new list on each outer iteration, resulting in a nested (2D) list.
[[i*j for i in range(1,4)] for j in range(1,4)]
```

```python
# cartesian products -> List comprehension on 2 lists together
L1 = [1,2,3,4]
L2 = [5,6,7,8]

# One pair of [] → One list is created.

[i*j for i in L1 for j in L2]
```

## Zip

The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
# Write a program to add items of 2 lists indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

print(list(zip(L1,L2)))

# [i+j for i,j in zip(L1,L2)]
```
